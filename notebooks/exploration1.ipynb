```python
import os

# Define the target directory path
folder = '../data/raw'

# 1. Scan the directory and pull all valid files
files = sorted([f for f in os.listdir(folder) if os.path.isfile(os.path.join(folder, f))])

# 2. Isolate unique ENCODE accession prefixes
accessions = sorted(list(set([f.split('.')[0] for f in files if f.startswith('ENCFF')])))

# 3. Consolidate unique file extensions
extensions = sorted(list(set([f.split('.')[-1] for f in files if '.' in f])))

# 4. Sum up total file weights and convert raw bytes to Gigabytes
total_bytes = sum(os.path.getsize(os.path.join(folder, f)) for f in files)
total_size_gb = total_bytes / (1024**3)

# 5. Print out the calculated details
print(f"Total files checked: {len(files)}")
print(f"Unique accessions: {accessions}")
print(f"File formats present: {extensions}")
print(f"Total dataset footprint: {total_size_gb:.2f} GB")
```
